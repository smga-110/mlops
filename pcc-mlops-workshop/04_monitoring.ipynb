# 04 · Inference Monitoring & Data Drift *(placeholder)*

**Team: ML Engineering / Deployment (Workspace B)**

Once the endpoint from `03` is live, Databricks can monitor it with almost no extra code. This
notebook is a **placeholder** we'll flesh out in a later session — the outline below shows where
we're headed.

### 1. Turn on inference logging on the endpoint
Enable an **inference table** so every request/response is durably logged to Unity Catalog. This
is a checkbox/config on the serving endpoint (AI Gateway `inference_table_config`) — no app code.

```
<catalog>.<schema>.<endpoint>_payload   # auto-populated request/response log (async, ~minutes)
```

### 2. Lakehouse Monitoring on the inference table
Point **Databricks Lakehouse Monitoring** (`InferenceLog` profile) at the parsed payload table:
- `prediction_col` = the model's output probability
- `problem_type` = classification
- `granularity` = e.g. 1 hour / 1 day

It auto-generates **profile** and **drift** metric tables and a **quality dashboard**.

### 3. Drift → retraining trigger
Wire a **Databricks Job** (retraining = notebook `01`) to fire when a drift metric breaches a
threshold, closing the MLOps loop: **train → deploy → monitor → retrain**.

### 4. AI/BI dashboard
A Lakeview dashboard over the payload + drift tables for the business view (volume, score
distribution, feature drift, risk bands).
